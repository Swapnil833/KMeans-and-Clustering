```python
import pandas as pd

X = pd.DataFrame(iris.data)

X
```

```python
X.columns = iris.feature_names

X.head()
```

```python
#Using them as Numpy arrays

X = iris.data
Y = iris.target
```

```python
sns.scatterplot(x = X[:, 0],  y = X[:, 1],  c=Y)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
# #optional - Dimentionality Reduction using PCA

# from sklearn.decomposition import PCA

# pca = PCA(n_components = 2)

# pca_data = pca.fit_transform(X_scaled)
```

```python
wcss = []

for k in range(1, 11):
    kmeans = KMeans(n_clusters = k)
    kmeans.fit(X_scaled)
    wcss.append(kmeans.inertia_)
```

```python
sns.lineplot(x = range(1, 11), y = wcss, marker = 'o')
```

```python
kmeans  = KMeans(n_clusters = 3)
labels = kmeans.fit_predict(X_scaled)

sns.scatterplot(x = X[:, 0], y = X[:, 2],  c = labels)
```

```python
sns.scatterplot(x = X[:, 0], y = X[:, 2],  c = Y)

#here , we are comparing kaise humara KMeans k detected labels aur humare original Y are same.
```

```python
#optional - Dimentionality Reduction using PCA

from sklearn.decomposition import PCA

pca = PCA(n_components = 2)

pca_data = pca.fit_transform(X_scaled)
```

```python
print(pca_data)
```

```python
wcss = []

for k in range(1, 11):
    kmeans = KMeans(n_clusters = k)
    kmeans.fit(pca_data)
    wcss.append(kmeans.inertia_)
```

```python
sns.lineplot(x = range(1, 11), y = wcss, marker = 'o')
```

```python
kmeans  = KMeans(n_clusters = 3)
labels = kmeans.fit_predict(pca_data)

sns.scatterplot(x = pca_data[:, 0], y = pca_data[:, 1],  c = labels)
sns.scatterplot(x = kmeans.cluster_centers_[:, 0], y = kmeans.cluster_centers_[:, 1], marker = "X", c="red", s = 200)
```

```python
#upar k are our PCA k clusters and niche ke are our original clusters

#basic inference hum ye nikal sakte hai ki PCA mai data is achese separated then the original data
```
